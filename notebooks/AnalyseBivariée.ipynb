```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

sns.countplot(data=df[df["Country"].isin(top10)], x="Country", hue="channel_type")
plt.title("Country vs channel_type")
plt.xticks(rotation=45)
plt.show()
```

interpretation:

- Relation entre les variables : Il existe une relation entre le pays et le type de chaîne, car la distribution des catégories varie selon les pays.

- Différences entre pays : India et United States ont les nombres de chaînes les plus élevés.

- Catégories dominantes : Les catégories Music et Entertainment sont parmi les plus fréquentes.

- Combinaisons rares : Certaines catégories comme Autos, Nonprofit ou Animals apparaissent très peu.

- Déséquilibre : Il existe un déséquilibre structurel, car la majorité des chaînes est concentrée dans quelques pays.

<h3>diargame en barre regrouper<h3>

```python
import pandas as pd
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

table.plot(kind="bar", figsize=(10,5))

plt.title("Country vs Channel Type")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

<h3>diagramme en berres empilées</h3>

```python
import pandas as pd
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

table.plot(kind="bar", stacked=True, figsize=(10,5))

plt.title("Country vs Channel Type")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

interpretation

- Relation entre variables : Il existe une relation entre le pays et le type de chaîne, car certains types apparaissent plus dans certains pays.

- Pays dominants : United States et India ont les nombres de chaînes les plus élevés.

- Catégories dominantes : Entertainment et Music sont les types de chaînes les plus fréquents.

- Combinaisons rares : Certaines catégories comme Animals, Nonprofit ou Autos apparaissent très peu dans plusieurs pays.

- Déséquilibre : La majorité des chaînes est concentrée dans quelques pays, surtout United States et India.

<h3>Heatmap du tableau de contingence</h3>

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

plt.figure(figsize=(10,6))
sns.heatmap(table, annot=True, cmap="YlGnBu")

plt.title("Heatmap: Country vs Channel Type")
plt.xlabel("Channel Type")
plt.ylabel("Country")
plt.show()
```

### Analyse bivaries des variables quantitative average yearly earning and video vues


```python
df['video views']
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = df['video views'] / 1_000_000_000
y = df['earnings_midpoint'] / 1_000_000


plt.figure(figsize=(10, 6))

hb = plt.hexbin(
    x, y,
    gridsize=22,
    cmap='Blues',
    mincnt=1,
    bins='log',
    edgecolors='black',
    linewidths=0.5
)

cb = plt.colorbar(hb)
cb.set_label('Nombre de points', fontsize=12)

plt.xlabel('Vues vidéo (en milliards)', fontsize=12)
plt.ylabel('Revenu annuel moyen (en millions)', fontsize=12)
plt.title('Impact des vues vidéo sur le revenu', fontsize=14, fontweight='bold')

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Paramétrage du graphique



-plus le nombre des youtuber augmente plus le couleur bleu devient plus foncé

-L’utilisation de bins='log' est pertinente, car elle améliore la lecture des différences de densité entre zones très peuplées et zones peu fréquentes.

-pour bonne presentation de l'axe x= nombre des vues on a choisie un millard par unité et pour axe y=revenue moyen en million vue que ceci concerne la totalité de l'annee.

## Interprétation


-Il existe une relation visible entre le revenu annuel moyen et le nombre de vues vidéo.

-La relation semble positive : lorsque le nombre de vues augmente, le revenu a globalement tendance à augmenter aussi.

-Cette relation paraît modérée, et non forte, car les points restent assez dispersés. Donc les vues influencent probablement le revenu, mais elles ne sont pas le seul facteur explicatif.

-On observe un groupe principal très dense dans la zone des faibles revenus et des faibles à moyennes vues. Cela suggère que la majorité des créateurs appartiennent à un profil “standard” ou “moyen”.


-Les combinaisons dominantes ou rares :
Les combinaisons les plus fréquentes sont situées en bas à gauche du graphique :
faible revenu + faible nombre de vues.
Les combinaisons rares concernent les très hauts revenus et les très grands volumes de vues.


Conclusion visuelle générale :
Le diagramme hexbin met en évidence une corrélation visuelle positive entre les vues vidéo et le revenu annuel moyen. Cependant, cette relation reste dispersée, ce qui signifie que d’autres variables peuvent aussi intervenir dans l’explication du revenu.
